```
import xcs_code_stockpile as xcs
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```
proposal_number = 101927227  # noqa
run = 47  # noqa
```

```
h5 = xcs.load_smalldata(run, proposal_number)
h5dict = xcs.H5Dictionary(
    stack='/jungfrau1M/ROI_area',
    i0=lambda h5: h5['diodeGon/channels'][..., 0],
    laser_on='evr/code_41',
    delay='scan/awg_delay_ns_delay'
)
```

```
metadata = xcs.create_metadata(h5, run, h5dict, roi=None)
valid = xcs.sensible_filter(metadata, 0.5, 99.5, 0.5, 99.5)
```

```
xcs.plot_detector_image(h5, h5dict, samplerate=100).show()
```

```
xcs.plot_i0_vs_time(metadata, valid).show()
```

```
xcs.plot_intensity_vs_i0(metadata, valid).show()
```

```
xcs.plot_intensity_vs_time(metadata, valid).show()
```

```
trxrd = xcs.calc_trxrd_stats(h5, h5dict, metadata, valid, 4169*2)
```

```


def normalize(a):
    return a / np.amax(a)


for i, tag in enumerate(['intensity_mean', 'com_i', 'com_j', 'fwhm_i', 'fwhm_j']):
    fig, ax = plt.subplots(1, 1)
    if i == 0:
        label_on = 'on'
        label_off = 'off'
        label_onoff = 'on/off'
    else:
        label_on = None
        label_off = None
        label_onoff = None
    ax.plot(trxrd.index, normalize(trxrd[f'on_{tag}']), label=label_on, color='red')
    ax.plot(trxrd.index, normalize(trxrd[f'off_{tag}']), label=label_off, color='blue')
    ax.plot(trxrd.index, normalize(trxrd[f'on_{tag}']/trxrd[f'off_{tag}']), label=label_onoff, color='black')
    ax.set_title(tag)
    if i == 0:
        ax.legend()
    fig.tight_layout()
    plt.show()
```
